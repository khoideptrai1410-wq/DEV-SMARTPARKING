# 09 — Model Comparison

Tổng hợp kết quả của ba bài toán trong Project 3. Mỗi bài toán được đánh giá bằng metric phù hợp, không so sánh Accuracy với MAE/RMSE trực tiếp.

```python
import os
import pandas as pd

classification_path = '../data/processed/classification_model_results.csv'
regression_path = '../data/processed/regression_model_results.csv'
demand_path = '../data/processed/demand_model_results.csv'

for path in [classification_path, regression_path, demand_path]:
    print(
        path,
        "→",
        "OK" if os.path.exists(path) else "THIẾU"
    )
```

```python
classification_results = pd.read_csv(classification_path)
regression_results = pd.read_csv(regression_path)
demand_results = pd.read_csv(demand_path)

print("Classification:")
display(classification_results.round(4))

print("\nRegression:")
display(regression_results.round(4))

print("\nDemand Prediction:")
display(demand_results.round(4))
```

```python
best_classification = (
    classification_results
    .sort_values(
        by=['macro_f1', 'accuracy'],
        ascending=[False, False]
    )
    .iloc[0]
)

best_regression = (
    regression_results
    .sort_values(
        by=['mae', 'rmse'],
        ascending=[True, True]
    )
    .iloc[0]
)

best_demand = (
    demand_results
    .sort_values(
        by=['mae', 'rmse'],
        ascending=[True, True]
    )
    .iloc[0]
)

summary = pd.DataFrame([
    {
        'task': 'Classification',
        'best_model': best_classification['model'],
        'primary_metric': 'Macro F1',
        'best_score': best_classification['macro_f1'],
        'secondary_metric': 'Accuracy',
        'secondary_score': best_classification['accuracy']
    },
    {
        'task': 'Regression',
        'best_model': best_regression['model'],
        'primary_metric': 'MAE',
        'best_score': best_regression['mae'],
        'secondary_metric': 'R²',
        'secondary_score': best_regression['r2']
    },
    {
        'task': 'Demand Prediction',
        'best_model': best_demand['model'],
        'primary_metric': 'MAE',
        'best_score': best_demand['mae'],
        'secondary_metric': 'R²',
        'secondary_score': best_demand['r2']
    }
])

display(summary.round(4))
```

```python
os.makedirs('../data/processed', exist_ok=True)

summary_path = '../data/processed/model_comparison_summary.csv'
summary.to_csv(summary_path, index=False)

print("Đã lưu:")
print(summary_path)

print("\nKết luận:")
for _, row in summary.iterrows():
    print(
        f"- {row['task']}: {row['best_model']} "
        f"({row['primary_metric']} = {row['best_score']:.4f})"
    )
```

## Quy tắc chọn model cuối

- Classification: ưu tiên **Macro F1** vì bài toán có nhiều lớp và không cân bằng.
- Regression: ưu tiên **MAE** vì dễ diễn giải sai số trung bình theo phút.
- Demand Prediction: ưu tiên **MAE** vì dễ diễn giải sai số trung bình theo số xe.
